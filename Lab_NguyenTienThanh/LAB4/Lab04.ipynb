This notebook uses the available seven Lab 2 command labels and adds the nine missing queue-announcement units with a local Vietnamese VITS checkpoint. The resulting inventory has 16 labels. Five augmented candidates are used for the original commands; two stochastic neural generations per added queue word provide alternatives.

Run `python prepare_neural_tts.py` once while connected, then inference runs with the checkpoint in local-only mode. The official Meta model is CC-BY-NC 4.0; do not use it commercially. No model weights are included in this project.


## Text normalization and inventory

Normalization lowercases, removes punctuation, and expands numbers digit by digit (`23` ? `hai ba`). Inventory metadata distinguishes Lab 2 audio from neural-generated units. The queue case is synthesized using the exact normalized phrase.


## Concatenation and unit selection

Baseline uses the first candidate and zero cross-fade. Dynamic programming minimizes duration/RMS target cost plus spectrum/energy join cost across all candidates; cross-fade sweeps 0/5/15/30 ms. Raw and selected queue outputs and boundary plots are saved.


## Prosody, HMM generation and neural backend

F0 and duration trials use librosa. The HMM demo generates five empirical state trajectories per unit and uses state duration to control a waveform hybrid; this is not a full parametric voice with vocoder. Meta MMS VITS generates the queue utterance and three matched test sentences; mel, waveform, inference latency and audio are saved.
